Driving eligibility


```python
age = int(input("Enter your age: "))

if age >= 18 and age < 60:
    print("Eligible to Drive")
else:
    print("Not Eligible")
```

NUMber check

```python
num = int(input("Enter a number: "))

if num == 0:
    print("Zero")
elif num > 0 and num % 2 == 0:
    print("Positive and Even")
elif num > 0 and num % 2 != 0:
    print("Positive and Odd")
else:
    print("Negative")
```

Voting eligibility

```python
age = int(input("Enter your age: "))
citizenship = input("Enter citizenship (Indian/Other): ")

if age >= 18 and citizenship == "Indian":
    print("Eligible to Vote")
else:
    print("Not Eligible")
```

Free entry

```python
age = int(input("Enter your age: "))
student = input("Are you a student? (Yes/No): ")

if age < 12 or student == "Yes":
    print("Free Entry")
else:
    print("Ticket Required")
```

Temperature Alert

```python
temp = float(input("Enter temperature: "))

if temp > 40:
    print("Extreme Heat")
elif temp >= 30:
    print("Hot")
elif temp >= 20:
    print("Normal")
else:
    print("Cold")
```
