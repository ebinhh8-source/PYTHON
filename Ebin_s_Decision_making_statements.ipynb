Write a program to find the smallest of three numbers.

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
num3 = int(input("Enter third number: "))

if num1 <= num2 and num1 <= num3:
    print("Smallest number is:", num1)
elif num2 <= num1 and num2 <= num3:
    print("Smallest number is:", num2)
else:
    print("Smallest number is:", num3)
```

Display the grade based on marks:

90–100 → A

80–89 → B

70–79 → C

60–69 → D

Below 60 → F



```python
marks = int(input("Enter marks: "))

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
else:
    grade = "F"

print("Grade:", grade)
```

Calculate electricity bill based on units consumed.
Slab rates
Units    
Rate

First 100 units
₹5/unit

101–200 units
₹7/unit

201–300 units
₹10/unit

Above 300 units
₹12/unit




```python
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + (units - 100) * 7
elif units <= 300:
    bill = (100 * 5) + (100 * 7) + (units - 200) * 10
else:
    bill = (100 * 5) + (100 * 7) + (100 * 10) + (units - 300) * 12

print("Electricity Bill = ₹", bill)
```

Calculate income tax based on different income slabs
Annual Income
Tax Rate

Up to ₹2,50,000
No tax

₹2,50,001 – ₹5,00,000
5%

₹5,00,001 – ₹10,00,000
20%

Above ₹10,00,000
30%




```python
income = int(input("Enter income: "))

if income <= 250000:
    tax = 0
elif income <= 500000:
    tax = income * 5 / 100
elif income <= 1000000:
    tax = income * 20 / 100
else:
    tax = income * 30 / 100

print("Income Tax =", tax)
```

Create a student grading system that accepts marks for 5 subjects and displays total, average, grade, and result.


```python
m1 = int(input("Mark 1: "))
m2 = int(input("Mark 2: "))
m3 = int(input("Mark 3: "))
m4 = int(input("Mark 4: "))
m5 = int(input("Mark 5: "))

total = m1 + m2 + m3 + m4 + m5
average = total / 5

if average >= 50:
    result = "Pass"
else:
    result = "Fail"

if average >= 90:
    grade = "A"
elif average >= 80:
    grade = "B"
elif average >= 70:
    grade = "C"
elif average >= 60:
    grade = "D"
else:
    grade = "F"

print("Total =", total)
print("Average =", average)
print("Grade =", grade)
print("Result =", result)

```
